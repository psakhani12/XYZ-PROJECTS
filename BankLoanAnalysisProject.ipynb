```python
#QUESTION 1

import pandas as pd

df = pd.read_excel(r"D:\loan_data.xlsx")

print("First 5 Records:")
print(df.head(5))

print("\nLast 5 Records:")
print(df.tail(5))
```

```python
#QUESTION 2

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("Column Names:")
print(df.columns)

print("Data Types:")
print(df.dtypes)
```

```python
#QUESTION 3

print("Missing Values:")
print(df.isnull().sum())
df = df.fillna(0)
print("Missing values")
```

```python
#QUESTION 4

print("Duplicate records:")
print(df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicates removed")
print(df.head())
```

```python
# QUESTION 5

df["Annual_Income"] = df["Monthly_Income"] * 12
print(df.head())
```

```python
# QUESTION 6

df["Loan_To_Income_Ratio"] = df["Loan_Amount"] / df["Annual_Income"]
print(df.head())
```

```python
# QUESTION 7

def risk(x):
    if x < 0.30:
        return "Low Risk"
    elif x <= 0.60:
        return "Medium Risk"
    else:
        return "High Risk"

df["Risk"] = df["Loan_To_Income_Ratio"].apply(risk)
print(df.head())
```

```python
# QUESTION 8

top10 = df.sort_values(by="Loan_Amount", ascending=False).head(10)
print("Top 10 Customers :")
print(top10)
```

```python
# QUESTION 9
loan = df.groupby("City")["Loan_Amount"].sum()
print("Total Loan AmounT:")
print(loan)
```

```python
# QUESTION 10

income = df.groupby("Occupation")["Monthly_Income"].mean()
print("Average Monthly Income:")
print(income)
```

```python
# QUESTION 11

occupation = df.groupby("Occupation").size()

print("Loan Applications:")
print(occupation)

print("Highest Applications:")
print(occupation.idxmax())
```

```python
# QUESTION 12

credit = df[df["Credit_Score"] < 600]
print("Score less than 600:")
print(credit)
```

```python
# QUESTION 13

risk = df.groupby("Risk_Level").size()
print("Risk Level:")
print(risk)
```

```python
# QUESTION 14

status = df.groupby("Loan_Status").size()
print(" Loan Applications:")
print(status)
```

```python
# QUESTION 15

import numpy as np

loan = np.array(df["Loan_Amount"])

print("Loan Amount:")
print("Mean:", np.mean(loan))
print("Median:", np.median(loan))
print("Maximum:", np.max(loan))
print("Minimum:", np.min(loan))
print("Standard Deviation:", np.std(loan))
```

```python
# QUESTION 16

df.to_excel("loan_data.xlsx", index=False)
print("Dataset exported")
```
